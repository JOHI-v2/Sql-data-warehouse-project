```sql
SELECT ROW_NUMBER() OVER(order BY customer_id),
       csi.cst_id customer_id,
       csi.cst_key customer_number,
       csi.cst_firstname customer_firstname,
       csi.cst_lastname customer_lastname,
       csi.cst_marital_status marital_status,
       csi.cst_create_date  create_date,
       csa.BDATE birthdate,
       loc.cntry country,
       CASE WHEN csi.cst_gndr = 'N/A' THEN COALESCE (csa.GEN, 'N/A') ELSE cst_gndr END AS gender
FROM   silver.crm_cust_info AS csi
       LEFT OUTER JOIN
       silver.erp_cust_az12 AS csa
       ON csi.cst_key = csa.CID
       LEFT JOIN Silver.erp_loc_a101 loc
       ON csi.cst_key = loc.CID ;
```

```sql
SELECT DISTINCT csi.cst_gndr,
                csa.GEN,
                CASE WHEN csi.cst_gndr = 'N/A' THEN COALESCE (csa.GEN, 'N/A') ELSE cst_gndr END AS gender
FROM   silver.crm_cust_info AS csi
       LEFT OUTER JOIN
       silver.erp_cust_az12 AS csa
       ON csi.cst_key = csa.CID;

SELECT *
FROM   Silver.crm_sales_details;
```

```sql
SELECT ROW_NUMBER() OVER (ORDER BY prd.cat_id, prd.prd_key) AS prd_key,
       prd.cat_id category_id,
       cat.CAT category,
       cat.SUBCAT subcategory,
       prd.prd_key product_number,
       prd.prd_nm product_name,
       prd.prd_cost cost,
       prd.prd_line "line",
       prd.prd_start_dt start_date,
       cat.MAINTENANCE maintenance
 from Silver.crm_prd_info prd
LEFT JOIN silver.erp_px_cat_g1v2 cat 
on prd.cat_id = cat.ID
WHERE prd_end_dt is NULL
```

```sql
SELECT TOP (1000) sls.sls_ord_num order_number,
                  cst.customer_key,
                  prd.product_key,
                  sls.sls_ord_dt order_date,
                  sls.sls_ship_dt ship_date,
                  sls.sls_due_dt due_date,
                  sls.sls_sales sales_amount,
                  sls.sls_quantity quantity,
                  sls.sls_price price
FROM   Silver.crm_sales_details AS sls
       LEFT OUTER JOIN
       gold.dim_customer AS cst
       ON sls.sls_cust_id = cst.customer_id
       LEFT OUTER JOIN
       gold.dim_product AS prd
       ON sls.sls_prd_key = prd.product_number
       ORDER BY sls_ord_num, customer_key, product_key
```
