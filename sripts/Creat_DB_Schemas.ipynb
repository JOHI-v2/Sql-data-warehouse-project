CREATE DATABASE AND SCHEMAS 
---------------------------
SCRIPT PURPOSE: 
    this script creates a Database named 'DataWarehouse' after checking if it esists and in that case it drops the existing one and recreate it, after, It creates there schemas for the Layers 'Bronze' , 'Silver' and 'Gold'



```sql
use master
```

```sql
IF EXISTS ( SELECT 1 FROM  sys.databases WHERE name = 'DataWarehouse')
BEGIN 
    ALTER DATABASE DataWarehouse
    SET SINGLE_USER WITH ROLLBACK IMMEDIATE;
    DROP DATABASE DataWarehouse
END;
    
```

```sql
Create DATABASE DataWarehouse
```

```sql
USE DataWarehouse  
```

```sql
CREATE SCHEMA Bronze;
```

```sql
CREATE SCHEMA Silver;
```

```sql
CREATE SCHEMA Gold;
```

```sql
SELECT
    name
FROM
    sys.schemas
```
